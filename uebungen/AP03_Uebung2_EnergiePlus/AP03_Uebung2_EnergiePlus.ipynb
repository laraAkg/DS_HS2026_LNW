## Setup (wird nicht bewertet)

````python
# Libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from sklearn import tree
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import (r2_score, mean_squared_error, accuracy_score, precision_score,
                             recall_score, confusion_matrix, RocCurveDisplay)

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Get current working directory
print(os.getcwd())
````

### Haushaltsdaten laden → `df`

````python
# Import data from csv to pandas dataframe named 'df'
df = pd.read_csv('household_data.csv')

# Show dimensions and first rows
print('Number of rows and columns:', df.shape)
df.head()
````

### Trainings- und Testdaten erstellen

Der folgende Code teilt die Daten **zweimal** in 80 % Trainingsdaten und 20 % Testdaten auf:

- **Verbrauch (Regression):** `X_train`, `X_test`, `y_train`, `y_test` mit den Features in `kwh_features` und der Zielvariable `annual_kwh`.
- **Zahlungsverzug (Klassifikation):** `X_train_class`, `X_test_class`, `y_train_class`, `y_test_class` mit den Features in `default_features` und der Zielvariable `payment_default`. `stratify` sorgt dafür, dass der Anteil der Zahlungsausfälle in Trainings- und Testdaten gleich gross ist.

````python
# Features and target for the consumption prediction (regression)
kwh_features = ['living_area_m2', 'persons', 'has_heatpump', 'has_ev', 'has_solar']
X_train, X_test, y_train, y_test = train_test_split(df[kwh_features],
                                                    df['annual_kwh'],
                                                    test_size=0.20,
                                                    random_state=42)

# Features and target for the payment default prediction (classification)
default_features = ['monthly_invoice_chf', 'past_late_payments', 'customer_years']
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(df[default_features],
                                                                            df['payment_default'],
                                                                            test_size=0.20,
                                                                            random_state=42,
                                                                            stratify=df['payment_default'])

# Show the size of the samples
print('Regression:     ', X_train.shape[0], 'Trainingsdaten /', X_test.shape[0], 'Testdaten')
print('Klassifikation: ', X_train_class.shape[0], 'Trainingsdaten /', X_test_class.shape[0], 'Testdaten')
````

### C3 (4 Punkte)

Erstellen Sie einen **Classification Tree** für `payment_default` mit einer maximalen Tiefe von 3 (`max_depth=3`, `random_state=42`) und trainieren Sie ihn mit `X_train_class` und `y_train_class`. Die ROC-Kurve mit der AUC auf den Testdaten ist vorgegeben.
