## Setup (wird nicht bewertet)

````python
# Libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from sklearn import tree
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import (r2_score, mean_squared_error, accuracy_score, precision_score,
                             recall_score, confusion_matrix, RocCurveDisplay)

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Get current working directory
print(os.getcwd())
````

### Haushaltsdaten laden → `df`

````python
# Import data from csv to pandas dataframe named 'df'
df = pd.read_csv('household_data.csv')

# Show dimensions and first rows
print('Number of rows and columns:', df.shape)
df.head()
````

### Trainings- und Testdaten erstellen

Der folgende Code teilt die Daten **zweimal** in 80 % Trainingsdaten und 20 % Testdaten auf:

- **Verbrauch (Regression):** `X_train`, `X_test`, `y_train`, `y_test` mit den Features in `kwh_features` und der Zielvariable `annual_kwh`.
- **Zahlungsverzug (Klassifikation):** `X_train_class`, `X_test_class`, `y_train_class`, `y_test_class` mit den Features in `default_features` und der Zielvariable `payment_default`. `stratify` sorgt dafür, dass der Anteil der Zahlungsausfälle in Trainings- und Testdaten gleich gross ist.

````python
# Features and target for the consumption prediction (regression)
kwh_features = ['living_area_m2', 'persons', 'has_heatpump', 'has_ev', 'has_solar']
X_train, X_test, y_train, y_test = train_test_split(df[kwh_features],
                                                    df['annual_kwh'],
                                                    test_size=0.20,
                                                    random_state=42)

# Features and target for the payment default prediction (classification)
default_features = ['monthly_invoice_chf', 'past_late_payments', 'customer_years']
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(df[default_features],
                                                                            df['payment_default'],
                                                                            test_size=0.20,
                                                                            random_state=42,
                                                                            stratify=df['payment_default'])

# Show the size of the samples
print('Regression:     ', X_train.shape[0], 'Trainingsdaten /', X_test.shape[0], 'Testdaten')
print('Klassifikation: ', X_train_class.shape[0], 'Trainingsdaten /', X_test_class.shape[0], 'Testdaten')
````

**Aufgabe:** Schreiben Sie den korrigierten Code in die Zelle «Korrigierter Code» (die Kommentare des ursprünglichen Codes sind als Gerüst vorgegeben). Begründen Sie jede Änderung gegenüber dem ursprünglichen Code und erklären Sie, warum das ursprüngliche Ergebnis zu optimistisch ist.

**Begründung der Änderungen:**

1. **Zielvariable aus den Features entfernen:** `payment_default` ist die Zielvariable und darf nicht gleichzeitig Feature sein. Das Modell «sieht» sonst die richtige Antwort und muss sie nur abschreiben. Zudem ist der Zahlungsverzug zum Zeitpunkt der Vorhersage noch gar nicht bekannt.
2. **Auswertung auf den Testdaten:** Vorhersage und Kennzahlen wurden mit den **Trainings**daten berechnet. Das erkennt man auch an der Confusion Matrix: 748 + 212 = **960** Haushalte, also die 80 % Trainingsdaten. Diese Haushalte kennt das Modell bereits, der Random Forest hat sie «auswendig gelernt» (Overfitting). Für neue Kunden sagt das nichts aus. Richtig: `predict(X_test_B2)` und Vergleich mit `y_test_B2`.

````python
# Create training and test data (without the target variable 'payment_default' as feature)
X_train_B2, X_test_B2, y_train_B2, y_test_B2 = train_test_split(df[['monthly_invoice_chf', 'past_late_payments', 'customer_years']],
                                                                df['payment_default'],
                                                                test_size=0.20,
                                                                random_state=42,
                                                                stratify=df['payment_default'])

# Create and train the random forest classifier
rfc_B2 = RandomForestClassifier(n_estimators=100, random_state=42)
rfc_B2.fit(X_train_B2, y_train_B2)

# Predict the target variable for the TEST data
y_pred_B2 = rfc_B2.predict(X_test_B2)

# Evaluate the model on the TEST data
print('Accuracy:', round(accuracy_score(y_test_B2, y_pred_B2), 3))
print('Confusion matrix:')
print(confusion_matrix(y_test_B2, y_pred_B2))
````

### D5 – Optimale Tiefe finden

Trainieren Sie Classification Trees für `payment_default` mit `max_depth` = 1, 2, …, 10 (`random_state=42`) und berechnen Sie jeweils die Accuracy auf **Trainings-** und **Testdaten**.

**Frage:** Welche Tiefe würden Sie wählen? Was passiert bei grosser Tiefe?

**Antwort auf die Frage:**

Die **Trainings-Accuracy steigt** mit jeder zusätzlichen Tiefe (bis 0.989), die **Test-Accuracy ist bei Tiefe 2 am höchsten (0.875)** und sinkt danach tendenziell (0.838 bei Tiefe 10). Das ist **Overfitting**. Empfehlung: **`max_depth=2`**.

**Python-Code:**

````python
# Lists for the results
depths, acc_train, acc_test = range(1, 11), [], []

for d in depths:
    # Create and train a classification tree with max_depth=d and random_state=42
    clf = DecisionTreeClassifier(max_depth=d, random_state=42)
    clf.fit(X_train_class, y_train_class)
    # Accuracy on training and test data
    acc_train.append(accuracy_score(y_train_class, clf.predict(X_train_class)))
    acc_test.append(accuracy_score(y_test_class, clf.predict(X_test_class)))

# Plot (given)
plt.figure(figsize=(7, 4))
plt.plot(depths, acc_train, marker='o', label='Trainingsdaten')
plt.plot(depths, acc_test, marker='o', label='Testdaten')
plt.xlabel('max_depth'); plt.ylabel('Accuracy'); plt.title('Accuracy vs. Tiefe des Trees')
plt.legend(); plt.grid()
plt.show()
````
