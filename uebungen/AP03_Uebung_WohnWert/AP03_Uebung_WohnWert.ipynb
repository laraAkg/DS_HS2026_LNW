### Trainings- und Testdaten erstellen

Der folgende Code teilt die Daten **zweimal** in 80 % Trainingsdaten und 20 % Testdaten auf:

- **Miete (Regression):** `X_train`, `X_test`, `y_train`, `y_test` mit den Features in `rent_features` und der Zielvariable `rent_chf`.
- **Kündigung (Klassifikation):** `X_train_class`, `X_test_class`, `y_train_class`, `y_test_class` mit den Features in `moveout_features` und der Zielvariable `tenant_moves_out`. `stratify` sorgt dafür, dass der Anteil der Kündigungen in Trainings- und Testdaten gleich gross ist.

````python
# Features and target for the rent prediction (regression)
rent_features = ['area_m2', 'distance_center_km', 'building_age_years', 'has_balcony', 'floor']
X_train, X_test, y_train, y_test = train_test_split(df[rent_features],
                                                    df['rent_chf'],
                                                    test_size=0.20,
                                                    random_state=42)

# Features and target for the move-out prediction (classification)
moveout_features = ['rent_increase_pct', 'complaints_count', 'tenancy_years']
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(df[moveout_features],
                                                                            df['tenant_moves_out'],
                                                                            test_size=0.20,
                                                                            random_state=42,
                                                                            stratify=df['tenant_moves_out'])

# Show the size of the samples
print('Regression:     ', X_train.shape[0], 'Trainingsdaten /', X_test.shape[0], 'Testdaten')
print('Klassifikation: ', X_train_class.shape[0], 'Trainingsdaten /', X_test_class.shape[0], 'Testdaten')
````

Accuracy = **0.867**. Ein Modell, das immer «bleibt» sagt, erreicht bereits 183 / 240 = **0.763**. Der Random Forest ist also nur rund 10 Prozentpunkte besser. Die Accuracy allein ist wegen der ungleichen Klassen wenig aussagekräftig.

Recall (Klasse 1) = TP / (TP + FN) = 33 / (33 + 24) = **0.58**: Gut 4 von 10 kündigenden Mietern werden übersehen.

AUC = **0.86**: Das Modell kann kündigende und bleibende Mieter **gut** unterscheiden (0.5 = Raten, 1.0 = perfekt). Die AUC bewertet alle möglichen Schwellenwerte. Mit einem tieferen Schwellenwert als 0.5 liesse sich der Recall erhöhen (vgl. D2).
